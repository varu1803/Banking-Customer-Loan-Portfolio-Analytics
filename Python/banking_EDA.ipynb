```python
# Validate banking business rules

checks = {
    "Customers below 18": (customers["age"] < 18).sum(),
    "Customers above 100": (customers["age"] > 100).sum(),
    "Negative annual income": (customers["annual_income_inr"] < 0).sum(),
    "Credit score outside 300-900": (
        ~customers["credit_score"].between(300, 900)
    ).sum(),

    "Negative account balance": (
        accounts["current_balance_inr"] < 0
    ).sum(),

    "Non-positive loan amount": (
        loans["loan_amount_inr"] <= 0
    ).sum(),
    "Invalid interest rate": (
        (loans["annual_interest_rate_pct"] <= 0) |
        (loans["annual_interest_rate_pct"] > 50)
    ).sum(),
    "Non-positive loan tenure": (
        loans["tenure_months"] <= 0
    ).sum(),
    "Negative disbursed amount": (
        loans["disbursed_amount_inr"] < 0
    ).sum(),

    "Non-positive EMI due": (
        payments["emi_due_inr"] <= 0
    ).sum(),
    "Negative payment amount": (
        payments["amount_paid_inr"] < 0
    ).sum(),
    "Negative days past due": (
        payments["days_past_due"] < 0
    ).sum()
}

for check, count in checks.items():
    print(f"{check}: {count}")
```

```python
# Loan status and disbursement validation

print("Rejected loans with positive disbursement:",
      (
          (loans["loan_status"] == "Rejected") &
          (loans["disbursed_amount_inr"] > 0)
      ).sum())

print("Non-rejected loans with zero disbursement:",
      (
          (loans["loan_status"] != "Rejected") &
          (loans["disbursed_amount_inr"] <= 0)
      ).sum())

print("Disbursed amount greater than loan amount:",
      (
          loans["disbursed_amount_inr"] >
          loans["loan_amount_inr"]
      ).sum())
```

```python
# Extract rejected loans

rejected_loans = loans[
    loans["loan_status"] == "Rejected"].copy()

print("Rejected Loans:", len(rejected_loans))

# Export rejected loans separately
rejected_loans.to_csv(
    "Cleaned_Data/rejected_loans.csv",
    index=False,
    na_rep="\\N"
)

print("Rejected loans CSV created successfully!")
```

```python
# Generate SQL INSERT statements for rejected loans

def sql_value(value):
    if pd.isna(value):
        return "NULL"
    if isinstance(value, pd.Timestamp):
        return "'" + value.strftime("%Y-%m-%d") + "'"
    if isinstance(value, str):
        return "'" + value.replace("\\", "\\\\").replace("'", "''") + "'"
    return str(value)

columns = [
    "loan_id",
    "customer_id",
    "loan_type",
    "loan_amount_inr",
    "annual_interest_rate_pct",
    "tenure_months",
    "application_date",
    "disbursement_date",
    "loan_status",
    "disbursed_amount_inr"
]

with open(
    "Cleaned_Data/insert_rejected_loans.sql",
    "w",
    encoding="utf-8"
) as f:

    f.write("USE banking_analytics_db;\n\n")

    for _, row in rejected_loans.iterrows():
        values = ", ".join(
            sql_value(row[col]) for col in columns
        )

        query = (
            f"INSERT INTO bank_loans "
            f"({', '.join(columns)}) "
            f"VALUES ({values});\n"
        )

        f.write(query)

print("SQL file generated successfully!")
print("Total INSERT statements:", len(rejected_loans))
```

```python
# Generate SQL INSERT statements for payments

payment_columns = [
    "payment_id",
    "loan_id",
    "due_date",
    "payment_date",
    "emi_due_inr",
    "amount_paid_inr",
    "payment_status",
    "days_past_due"
]

with open(
    "Cleaned_Data/insert_payments.sql",
    "w",
    encoding="utf-8"
) as f:

    f.write("USE banking_analytics_db;\n\n")

    for _, row in payments.iterrows():

        values = ", ".join(
            sql_value(row[col]) for col in payment_columns
        )

        query = (
            f"INSERT INTO loan_payments "
            f"({', '.join(payment_columns)}) "
            f"VALUES ({values});\n"
        )

        f.write(query)

print("SQL file generated successfully!")
print("Total INSERT statements:", len(payments))
```
